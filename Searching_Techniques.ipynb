### Linear Search

Linear search is the simplest searching algorithm. It sequentially checks each element of the list until a match is found or the whole list has been searched. It's suitable for small or unsorted lists.

```python
def linear_search(arr, target):

    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Example usage of linear search
my_list_linear = [4, 2, 7, 1, 9, 5]
target_linear = 9

index_linear = linear_search(my_list_linear, target_linear)

if index_linear != -1:
    print(f"Linear Search: Element {target_linear} found at index {index_linear}")
else:
    print(f"Linear Search: Element {target_linear} not found in the list")

target_linear_not_found = 3
index_linear_not_found = linear_search(my_list_linear, target_linear_not_found)
if index_linear_not_found != -1:
    print(f"Linear Search: Element {target_linear_not_found} found at index {index_linear_not_found}")
else:
    print(f"Linear Search: Element {target_linear_not_found} not found in the list")
```

### Binary Search

Binary search is an efficient algorithm for finding an item from a *sorted* list of items. It works by repeatedly dividing in half the portion of the list that could contain the item, until you've narrowed down the possible locations to just one. Binary search is significantly faster than linear search for large sorted lists.

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        mid_element = arr[mid]

        if mid_element == target:
            return mid
        elif mid_element < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Example usage of binary search (list MUST be sorted)
my_list_binary = [1, 2, 4, 5, 7, 9]
target_binary = 7

index_binary = binary_search(my_list_binary, target_binary)

if index_binary != -1:
    print(f"Binary Search: Element {target_binary} found at index {index_binary}")
else:
    print(f"Binary Search: Element {target_binary} not found in the list")

target_binary_not_found = 6
index_binary_not_found = binary_search(my_list_binary, target_binary_not_found)
if index_binary_not_found != -1:
    print(f"Binary Search: Element {target_binary_not_found} found at index {index_binary_not_found}")
else:
    print(f"Binary Search: Element {target_binary_not_found} not found in the list")
```
