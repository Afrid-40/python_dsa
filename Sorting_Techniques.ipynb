# **`Bubble Sort`**

```python
a =[ 5,3,2,1,4]
# bubble sort repeatedly compares with adjacent elements.
for i in range(len(a)):
    for j in range(len(a)-i-1):
        if a[j] > a[j+1]:
            a[j],a[j+1] = a[j+1],a[j]
print(a)
```

# **`Selection Sort`**

```python
a = [1, 5, 8, 0, 2, 4]
# Selection sort finds the smallest element and places it at the correct position
for i in range(len(a)):
    m = i
    for j in range(i+1,len(a)):
        if a[j]<a[m]:
            m=j
    a[i],a[m] = a[m],a[i]
print(a)
```

# **`Insertion Sort`**

```python
a = [1,5,8,0,2,4]
#insertion sort builds the sorted portion one element at a time
for i in range(1,len(a)):
    k=a[i]
    j = i - 1
    while j >= 0 and a[j]>k:
        a[j + 1] = a[j]
        j-=1
        a[j + 1] = k
print(a)
```

# **`Merge Sort`**

```python
a=[2,5,6,1,3,4]
# merge sort builds the sort portions by using divide and conquer method.
def merge_sort(a):
    if len(a) <=1:
        return a
    mid = len(a) //2
    left= merge_sort(a[:mid])
    right = merge_sort(a[mid:])
    r=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            r.append(left[i])
            i +=1
        else:
            r.append(right[j])
            j +=1
    r.extend(left[i:])
    r.extend(right[j:])
    return r
merge_sort(a)
```

# **`Quick-Sort `**

```python
a = [8,3,5,4,7,6,1,2]
def quick(a):
    if len(a) <=1:
        return a
    pivot = a[-1]
    left = []
    right =[]
    for i in range(len(a)-1):
        if (a[i]<pivot):
            left.append(a[i])
        else:
            right.append(a[i])
    return quick(left)+[pivot] + quick(right)
print(quick(a))
```
