```python
# find how many vowel , consonants, spaces and specail characters
vowels = 'aeiou'
x = "python is a programming language"
vowel_count = 0
consonant_count = 0
space_count = 0
special_char_count = 0
x_small = x.lower()

for char in x_small:
    if char in vowels:
        vowel_count += 1
    elif char.isalpha():
        consonant_count += 1
    elif char.isspace():
        space_count += 1
    else:
        special_char_count += 1
print(vowel_count)
print(consonant_count)
print(space_count)
print(special_char_count)
```

```python
def build_prefix(arr):
  prefix=[arr[0]]
  for i in range(1,len(arr)):
    prefix.append(prefix[i-1]+arr[i])
  return prefix
```

```python
#equillibrium index
n=[-7,1,5,2,-4,3,0]
def equillibrium_index(n):
  left_sum=0
  total=sum(n)
  for i in range(len(n)):
    right_sum = total-n[i]-left_sum
    if left_sum == right_sum:
      return i
    left_sum += n[i]
  return -1
print(equillibrium_index(n))
```

```python
# subarrray equals to total k
arr2=[1,2,3,4]
k = 7
def subarray_sum_equals_k(arr, target_sum):
    current_sum = 0
    prefix_sums = {0}
    for num in arr:
        current_sum += num
        if (current_sum - target_sum) in prefix_sums:
            return True
        prefix_sums.add(current_sum)
    return False
result = subarray_sum_equals_k(arr2, k)
print(result)
```

```python
arr2=[1,2,3,4]
k=7
def subarray_sum(arr,k):
  current_sum =0
  start = 0
  for end in range(len(arr)):
    current_sum +=arr[end]
    while current_sum > k and start < end:
      current_sum -= arr[start]
      start += 1
    if current_sum == k:
      return start, end
  return -1
print(subarray_sum(arr2,k))
```
