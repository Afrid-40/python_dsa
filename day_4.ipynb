```python
# string manipulation
vowels= 'aeiou'
x="python is a programming language"
count = 0
x_small = x.lower()
for x in x_small:
    if x in vowels:
        count += 1
print(count)
```

```python
# find how many vowel , consonants, spaces and specail characters
vowels = 'aeiou'
x = "python is a programming language"
vowel_count = 0
consonant_count = 0
space_count = 0
special_char_count = 0
x_small = x.lower()

for char in x_small:
    if char in vowels:
        vowel_count += 1
    elif char.isalpha():
        consonant_count += 1
    elif char.isspace():
        space_count += 1
    else:
        special_char_count += 1
print(vowel_count)
print(consonant_count)
print(space_count)
print(special_char_count)
```

```python
#get input from the user as string
# remove duplicates characters from the string
x= input("enter the string")
s = set()
result = ""
for i in x:
  if i in s:
    continue
  s.add(i)
  result += i
print(result)

```

```python
# get input from user reverse it
#use for loop no slicing [::1] dont user
x = input()
rev = ""
for i in x:
  rev = i+rev
print(rev)
```

```python
# Anagram eg: listen and silent
# write code to find anagram
s = input()
s1= input()
# print(sorted(s)==sorted(s1))
if len(s) == len(s1):
  print(False)
frq={}
for ch in s:
  frq[ch] = frq.get(ch,0)+1
for ch in s1:
  if ch not in frq:
    print(False)
    break
  else:
    frq[ch] = frq[ch]-1
else:
  print(True)

```

```python
#recursion
def fact(n):
  if n<= 1:
    return 1
  return n*fact(n-1)
fact(8)
```

```python
#fibnoacci
def fib(n):
  if n<=1:
    return n
  return fib(n-1) + fib(n-2)
for i in range (5):
  print(fib(i),end=" ")
```
