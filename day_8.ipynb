```python
class TreeNode:
  def __init__(self,data): # Corrected _init_ to __init__
    self.data=data
    self.children=[]

class Tree:
  def __init__(self): # Corrected _init_ to __init__
    self.root=None

  def add(self,data,parentdata=None): # Moved inside Tree class
    node=TreeNode(data)
    if not self.root:
      self.root=node
      return
    #to find where new data is added
    #which parent node to be added
    parentNode=self.findNode(parentdata,self.root)
    #if parentnode found then add node
    if parentNode:
      parentNode.children.append(node)
    else:
      print(f"Parent node {parentdata} not found")

  def findNode(self,data,node): # Moved inside Tree class
    if node.data==data:
      return node
    #recursively check for which parent node
    for child in node.children:
      nodeFound=self.findNode(data,child)
      if nodeFound:
        return nodeFound
    #if node not found return None
    return None

  def display(self,depth=0,node=None): # Moved inside Tree class
    if not node:
      node=self.root
    print(" "*depth,node.data)
    for child in node.children:
      self.display(depth+1,child)
```

```python
from collections import deque

# Adding traversal methods to the Tree class
class Tree:
  def __init__(self):
    self.root=None

  def add(self,data,parentdata=None):
    node=TreeNode(data)
    if not self.root:
      self.root=node
      return
    parentNode=self.findNode(parentdata,self.root)
    if parentNode:
      parentNode.children.append(node)
    else:
      print(f"Parent node {parentdata} not found")

  def findNode(self,data,node):
    if not node: # Handle case where node is None, which can happen in recursive calls if not careful
      return None
    if node.data==data:
      return node
    for child in node.children:
      nodeFound=self.findNode(data,child)
      if nodeFound:
        return nodeFound
    return None

  def display(self,depth=0,node=None):
    if not node:
      node=self.root
    if not node: # Handle empty tree case for display
      print("Tree is empty")
      return
    print(" "*depth + str(node.data))
    for child in node.children:
      self.display(depth+1,child)

  def pre_order_traversal(self, node=None):
    if not node:
      node = self.root
    if not node: # Handle empty tree
      return
    print(node.data, end=" ")
    for child in node.children:
      self.pre_order_traversal(child)

  def post_order_traversal(self, node=None):
    if not node:
      node = self.root
    if not node: # Handle empty tree
      return
    for child in node.children:
      self.post_order_traversal(child)
    print(node.data, end=" ")

  def level_order_traversal(self):
    if not self.root:
      return

    queue = deque([self.root])
    while queue:
      node = queue.popleft()
      print(node.data, end=" ")
      for child in node.children:
        queue.append(child)
```

```python
# Create a sample tree
tree = Tree()
tree.add("A")
tree.add("B", "A")
tree.add("C", "A")
tree.add("D", "B")
tree.add("E", "B")
tree.add("F", "C")

print("Tree Structure:")
tree.display()
```

```python
# Pre-order Traversal (Root -> Left -> Right for binary, Root -> Children for N-ary)
print("\nPre-order Traversal:")
tree.pre_order_traversal()
print()
```

```python
# Post-order Traversal (Left -> Right -> Root for binary, Children -> Root for N-ary)
print("\nPost-order Traversal:")
tree.post_order_traversal()
print()
```

```python
# Level-order Traversal (Breadth-First Search)
print("\nLevel-order Traversal:")
tree.level_order_traversal()
print()
```
