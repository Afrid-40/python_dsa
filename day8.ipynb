```python
#Doubly LinkedList
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None
    self.prev = None
class DoublyLinkedList:
  def __init__(self):
    self.head = None
  def add_at_end(self, data):
    new_node = Node(data)
    if not self.head:
      self.head = new_node
    else:
      curr = self.head
      while curr.next:
        curr = curr.next
      curr.next = new_node
      new_node.prev = curr
  def add_at_start(self,data):
    new_node = Node(data)
    if not self.head:
      self.head = new_node
    else:
      self.head.prev = new_node
      new_node.next = self.head
      self.head = new_node
  #[None|5|10 address <-->5 address|10|20 address <--> 10 address|20|None]
  def display(self):
    curr = self.head
    print( "None ", end="")
    while curr is not None:
      print(curr.data, end = " ")
      curr = curr.next
    print(" None")
  def remove_at_end(self):
    if not self.head: # List is empty
      return
    elif not self.head.next: # Only one node in the list
      self.head = None
      return
    else:
      curr = self.head
      while curr.next.next: # Traverse to the second to last node
        curr = curr.next
      # At this point, 'curr' is the second-to-last node.
      curr.next = None # Remove the last node by breaking the link
ddl=DoublyLinkedList()
ddl.add_at_end(5)
ddl.add_at_end(10)
ddl.add_at_start(5)
ddl.add_at_start(10)
ddl.display()
ddl.remove_at_end()
ddl.display()
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None
class CircularLinkedList:
  def __init__(self):
    self.head = None
  def add_at_end(self, data):
    new_node = Node(data)
    if not self.head:
      self.head = new_node
      new_node.next = self.head
    else:
      curr = self.head
      while curr.next != self.head:
        curr = curr.next
      curr.next = new_node
      new_node.next = self.head
```
