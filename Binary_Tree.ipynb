```python
class BinaryNode:
  def __init__(self,data):
    self.data  = data
    self.left  = None
    self.right = None
```

```python
class BinaryTree:
  def __init__(self):
    self.root = None

  def add(self, data):
    node = BinaryNode(data)
    if self.root is None:
      self.root = node
      return

    # Use a queue for a level-order (BFS-like) insertion to find the first available spot.
    # This ensures a relatively balanced tree for arbitrary data insertion.
    q = [self.root]
    while q:
        current = q.pop(0)
        if current.left is None:
            current.left = node
            return
        else:
            q.append(current.left)
        if current.right is None:
            current.right = node
            return
        else:
            q.append(current.right)

  # In-order Traversal (Left -> Root -> Right)
  def in_order_traversal(self):
    elements = []
    self._in_order_recursive(self.root, elements)
    return elements

  def _in_order_recursive(self, node, elements):
    if node:
      self._in_order_recursive(node.left, elements)
      elements.append(node.data)
      self._in_order_recursive(node.right, elements)

  # Pre-order Traversal (Root -> Left -> Right)
  def pre_order_traversal(self):
    elements = []
    self._pre_order_recursive(self.root, elements)
    return elements

  def _pre_order_recursive(self, node, elements):
    if node:
      elements.append(node.data)
      self._pre_order_recursive(node.left, elements)
      self._pre_order_recursive(node.right, elements)

  # Post-order Traversal (Left -> Right -> Root)
  def post_order_traversal(self):
    elements = []
    self._post_order_recursive(self.root, elements)
    return elements

  def _post_order_recursive(self, node, elements):
    if node:
      self._post_order_recursive(node.left, elements)
      self._post_order_recursive(node.right, elements)
      elements.append(node.data)

  # Search for a node
  def search(self, target_data):
    return self._search_recursive(self.root, target_data)

  def _search_recursive(self, node, target_data):
    if node is None:
      return False
    if node.data == target_data:
      return True
    # For a general binary tree, we must search both subtrees
    return self._search_recursive(node.left, target_data) or self._search_recursive(node.right, target_data)

  # Get the height of the tree (number of edges on the longest path from root to a leaf)
  def get_height(self):
    return self._get_height_recursive(self.root)

  def _get_height_recursive(self, node):
    if node is None:
      return -1 # Height of an empty tree is -1
    left_height = self._get_height_recursive(node.left)
    right_height = self._get_height_recursive(node.right)
    return 1 + max(left_height, right_height)
```

```python
class BinarySearchTree:
  def __init__(self):
    self.root = None

  def add(self, data):
    if self.root is None:
      self.root = BinaryNode(data)
    else:
      self._add_recursive(self.root, data)

  def _add_recursive(self, current_node, data):
    if data < current_node.data:
      if current_node.left is None:
        current_node.left = BinaryNode(data)
      else:
        self._add_recursive(current_node.left, data)
    else: # data >= current_node.data, allow duplicates to go to the right
      if current_node.right is None:
        current_node.right = BinaryNode(data)
      else:
        self._add_recursive(current_node.right, data)

  # In-order Traversal (Left -> Root -> Right) - produces sorted elements for BST
  def in_order_traversal(self):
    elements = []
    self._in_order_recursive(self.root, elements)
    return elements

  def _in_order_recursive(self, node, elements):
    if node:
      self._in_order_recursive(node.left, elements)
      elements.append(node.data)
      self._in_order_recursive(node.right, elements)

  # Pre-order Traversal (Root -> Left -> Right)
  def pre_order_traversal(self):
    elements = []
    self._pre_order_recursive(self.root, elements)
    return elements

  def _pre_order_recursive(self, node, elements):
    if node:
      elements.append(node.data)
      self._pre_order_recursive(node.left, elements)
      self._pre_order_recursive(node.right, elements)

  # Post-order Traversal (Left -> Right -> Root)
  def post_order_traversal(self):
    elements = []
    self._post_order_recursive(self.root, elements)
    return elements

  def _post_order_recursive(self, node, elements):
    if node:
      self._post_order_recursive(node.left, elements)
      self._post_order_recursive(node.right, elements)
      elements.append(node.data)

  # Search for a node (BST-specific, more efficient)
  def search(self, target_data):
    return self._search_recursive(self.root, target_data)

  def _search_recursive(self, node, target_data):
    if node is None:
      return False
    if node.data == target_data:
      return True
    elif target_data < node.data:
      return self._search_recursive(node.left, target_data)
    else: # target_data > node.data
      return self._search_recursive(node.right, target_data)

  # Find the minimum value in the BST
  def find_min(self):
    if self.root is None:
      return None
    current = self.root
    while current.left:
      current = current.left
    return current.data

  # Find the maximum value in the BST
  def find_max(self):
    if self.root is None:
      return None
    current = self.root
    while current.right:
      current = current.right
    return current.data

  # Get the height of the tree (number of edges on the longest path from root to a leaf)
  def get_height(self):
    return self._get_height_recursive(self.root)

  def _get_height_recursive(self, node):
    if node is None:
      return -1 # Height of an empty tree is -1
    left_height = self._get_height_recursive(node.left)
    right_height = self._get_height_recursive(node.right)
    return 1 + max(left_height, right_height)



```
