```python
# stack Function
def fact(n):
  if n == 0 or n==1:
    return 1
  else:
    return n * fact(n-1)

print(fact(5))
```

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:
        self.stack.append(val)
        if self.min_stack:
            current_min = min(val, self.min_stack[-1])
        else:
            current_min = val
        self.min_stack.append(current_min)

    def pop(self) -> None:
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]
```
