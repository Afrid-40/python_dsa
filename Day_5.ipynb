```python
my_array = [3,4,2,1]

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        lp = 0
        rp = 0
        fp = 0

        while lp < len(left) and rp < len(right):
            if left[lp] < right[rp]:
                arr[fp] = left[lp]
                lp += 1
            else:
                arr[fp] = right[rp]
                rp += 1
            fp += 1

        while lp < len(left):
            arr[fp] = left[lp]
            lp += 1
            fp += 1

        while rp < len(right):
            arr[fp] = right[rp]
            rp += 1
            fp += 1

print("Before sorting:", my_array)
merge_sort(my_array)
print("After sorting:", my_array)
```

```python
import random
arr=[3,4,2,1,5,8,7]
def quick_sort(arr):
  if len(arr)<=1:
    return arr
  pivot=random.choice(arr)
  left=[i for i in arr if i<pivot]
  right=[i for i in arr if i > pivot]
  middle= [i for i in arr if i == pivot]
  return quick_sort(left)+middle+quick_sort(right)
print("Before Sorting:",arr)
print("After Sorting:", quick_sort(arr))
```
